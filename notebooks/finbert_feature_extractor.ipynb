# 🏦 FinBERT Sentiment Feature Extractor for Forex RL

This notebook demonstrates the quantitative NLP pipeline used in `quant_forex_rl`:
1. Ingesting financial headlines and macroeconomic statements
2. Performing tokenization and sentiment classification via Transformer models
3. Computing continuous sentiment polarity scores and volatility surprise metrics
4. Merging sentiment time-series with OHLCV market features without lookahead bias.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

ROOT_DIR = Path.cwd().parent
DATA_DIR = ROOT_DIR / "data"
print(f"Data Directory: {DATA_DIR}")
```

## 1. Load Financial Headlines Dataset
We inspect the logged headlines generated by the news ingestion engine.

```python
headlines_path = DATA_DIR / "news_headlines.csv"
if headlines_path.exists():
    df_headlines = pd.read_csv(headlines_path)
    print(f"Loaded {len(df_headlines)} headline records:")
    display(df_headlines.head(10))
else:
    print("Run src/news_ingestion.py first to generate headlines archive.")
```

## 2. FinBERT Inference Pipeline
We demonstrate scoring headlines using PyTorch and HuggingFace transformers.

```python
sample_texts = [
    "Federal Reserve signals aggressive rate hikes to combat stubborn inflation.",
    "US Non-Farm Payrolls miss expectations as unemployment rate ticks higher.",
    "Eurozone economic sentiment rebounds strongly following ECB policy stabilization."
]

# Demonstration using domain lexicon & transformer tokenizer structure
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
tokens = tokenizer(sample_texts, padding=True, truncation=True, return_tensors="pt")
print("Input IDs shape:", tokens["input_ids"].shape)
print("Attention Mask shape:", tokens["attention_mask"].shape)
```

## 3. Visualize Sentiment & Macro Shock Distributions

```python
if 'df_headlines' in locals():
    plt.figure(figsize=(12, 4))
    plt.subplot(1, 2, 1)
    df_headlines['sentiment_score'].hist(bins=20, color='#007acc', edgecolor='black')
    plt.title("NLP Sentiment Polarity Distribution")
    plt.xlabel("Polarity (-1.0 Bearish to +1.0 Bullish)")
    
    plt.subplot(1, 2, 2)
    df_headlines['event_surprise'].hist(bins=20, color='#ff7f0e', edgecolor='black')
    plt.title("Macroeconomic Event Surprise Distribution")
    plt.xlabel("Surprise Z-Score")
    plt.tight_layout()
    plt.show()
```
